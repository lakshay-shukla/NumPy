Numpy array vs Python lists

```python
# Speed
#1. List
a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]
c = []
import time
start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])
print(time.time()-start)

#2. Numpy Array
import numpy as np
a = np.arange(10000000)
b = np.arange(10000000,20000000)
start = time.time()
c = a + b
print(time.time()-start)

# Memory
#1. List
a = [i for i in range(10000000)]
import sys
sys.getsizeof(a)

#2. Numpy Array
b = np.arange(10000000)
sys.getsizeof(b)
```

Advanced Indexing
1. Normal Indexing and Slicing
2. Fancy Indexing
3. Boolean Indexing

```python
a = np.arange(12).reshape(4,3)
print(a)
#1. Normal Indexing
print(a[2,1]) # 7 (2 show = row no. and 3 show = column no.)

#2. Fancy Indexing --> When we can't find any pattern in normal indexing then we use Fancy indexing
print(a[[0,2,3]])  # we make a list and provide row no. [0,2,3]; 0 = row,2 = row,3 = row
print(a[:,[0,2]])  # This for Column Same as row
```

Boolean Indexing --> (Most imp) Using When we Fetch the item from giving condition.

Boolean Masking --> It gives value rather than true/false.

 example.  a[a>5]

 When you work with boolean values then we use bitwise operator(&,~)

```python
b = np.random.randint(1,100,24).reshape(24)   # 1,100 range of items nad 24 no. of items
print(b)

#1. Find All no. greater then 50
print(b>50)    # it gives only true false
print(b[b>50])  # Boolean masking it give values

#2. find out all even no.
print(b[b%2==0])

#3. find all no. greater than 50 and are even.
print(b[(b>50)&(b%2==0)])
```

Broadcasting --> The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.
The smaller array is “broadcast” across the larger array so that they have compatible shapes.

It is used in techniques called Vectorization.

Broadcasting Rules:
1. Make the two arrays have the same number of dimensions.
If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.
2. Make each dimension of the two arrays the same size.
If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
3. If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.

```python
#Example 1.
a = np.arange(12).reshape(4,3)
b = np.arange(3)
print(a)
print(b)
print(a+b)

#Example 2.
a = np.arange(3).reshape(1,3)
b = np.arange(4).reshape(4,1)
print(a)
print(b)
print(a + b)

#Example 3.
a = np.array([1])
# shape -> (1,1)
b = np.arange(4).reshape(2,2)
# shape -> (2,2)
print(a)
print(b)
print(a+b)
```
