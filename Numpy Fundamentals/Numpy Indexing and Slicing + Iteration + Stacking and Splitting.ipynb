Indexing and Slicing

Indexing --> Fetching single item

Slicing --> Fetching multiple items

1-D Array

2-D Array = 1-D + 1-D Array

3-D Array = 2-D + 2-D Array

```python
import numpy as np
a1 = np.arange(10)  # 1-D
a2 = np.arange(12).reshape(3,4)  # 2-D
a3 = np.arange(27).reshape(3,3,3)  # 3-D

# Indexing 1-D
print(a1[-1])  # 9
print(a1[0])  # 0

# Indexing 2-D
print(a2[1,2]) # 6 (1 show = row no. and 2 show = column no.)
print(a2[2,3]) # 11 (2 show = row no. and 3 show = column no.)

# Indexing 3-D
print(a3[1,0,1])  # 5 (1 show = 2-D array no. ,0 show = row no. and 2 show = column no.)
print(a3[0,1,0]) # 2 (0 show = 2-D array no. ,1 show = row no. and 0 show = column no.)

# Slicing 1-D
print(a1[2:5]) # 2,3,4
print(a1[2:5:2]) # 2,4

# Slicing 2-D
print(a2[0,: ]) # 0,1,2,3 (Row Fetch)
print(a2[:,2])  # 2,6,10 (Column Fetch)
print(a2[0::2,1::2])

# Slicing 3-D
print(a3[1])
print(a3[::2])
print(a3[0,1,:])
print(a3[1,:,1])
```

Iteration on Numpy Array

```python
for i in a1:
    print(i)  # All items

for i in a2:
    print(i)  # single-single row (1-D Array)

for i in a3:
    print(i) # 2-D array

for i in np.nditer(a3):
    print(i)  # print all item of 3-D array
```

Reshaping

```python
#1. Reshape --> it's convert the given array into another shape
b = np.arange(1,11).reshape(5,2)  # 5 rows and 2 columns
print(b)

#2. Transpose --> It's change rows into column and column into rows.
print(np.transpose(a2))  # 1st method of transpose
print(a2.T)  # 2nd method of transpose

#3. Ravel --> It's convert any dimension array into 1-D array.
print(a2.ravel())
print(a3.ravel())
```

Stacking --> Perform Addition on 2 or more arrays

There are two types of stacking

1. Horizontal Stacking
2. Vertical Stacking

The Shape of the array is same

```python
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
#1. Horizontal Stacking
print(np.hstack((a4,a5)))   # shape = (3,8)

#2. Vertical Stacking
print(np.vstack((a4,a5)))   # shape = (6,4)
```

Splitting --> Opposite of Stacking (breaking) ; and It's Cut Arrays in equal parts.

There are two types of Splitting

1. Horizontal Splitting
2. Vertical Splitting

Example --> College student data split into branches

```python
#1. Horizontal Splitting
print(np.hsplit(a4,2))

#2. Vertical Splitting
print(np.vsplit(a5,3))
```
