What is numpy?
 1. NumPy is the fundamental package for scientific computing in Python. It is a Python library that provides a multidimensional array object, various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays, including mathematical, logical, shape manipulation, sorting, selecting, I/O, discrete Fourier transforms, basic linear algebra, basic statistical operations, random simulation and much more.

2. At the core of the NumPy package, is the ndarray object. This encapsulates n-dimensional arrays of homogeneous data types.

3. Most important thing in numpy is numpy array.

Numpy Arrays Vs Python Sequences
1. NumPy arrays have a fixed size at creation, unlike Python lists (which can grow dynamically). Changing the size of an ndarray will create a new array and delete the original.

2. The elements in a NumPy array are all required to be of the same data type, and thus will be the same size in memory.

3. NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python’s built-in sequences.

4. A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typically support Python-sequence input, they convert such input to NumPy arrays prior to processing, and they often output NumPy arrays.

```python
# Creating Numpy Arrays --> This is used to create arrays.
#1. np.array
import numpy as np
a = np.array([1,2,3])
print(a)  # [1 2 3]  1-D Array (Vector)

#2. 2D and 3D
b = np.array([[1,2,3],[4,5,6]])
print(b)   # 2-D Array (Matrix)
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(c)  # 3-D Array (Tensor)

#3. dtype --> data type array
d = np.array([1,2,3],dtype=float)
print(d)

#4. np.arange --> 2nd method to create array and it's like range function in 'for loop'.
e = np.arange(1,11,2)
print(e)
# with reshape --> it's convert the given array into another shape
f = np.arange(1,11).reshape(5,2)  # 5 rows and 2 columns
print(f)
# rows*column (5*2) should be equal to the item that are present inside the array(1,11).
```

```python
# Array Initialization Techniques
#1. np.ones and np.zeros
a = np.ones((3,4)) # we can make array which has all items are 1.
b = np.zeros((3,4)) # we can make array which has all items are 0.
print(a)
print(b)

#2. np.random --> it's generate random numbers array between 0 and 1.
c = np.random.random((3,4))
print(c)

#3. np.linspace --> (linear space)it's generate linearly separable point in given range.
d = np.linspace(-10,10,10,dtype=int) #-10=lower range,10=upper range,10=number of item
print(d) # in output difference (-) between 2 point\items are equal.

#4. np.identity --> it is used to create identity matrix.
e = np.identity(3)
print(e)
```

```python
# Numpy\Array Attributes
a1 = np.arange(10)
a2 = np.arange(12,dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)

#1. ndim --> number of dimensions
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)

#2. shape --> How many item in the array in the form of rows and columns
print(a1.shape)
print(a2.shape)
print(a3.shape)

#3. size --> Total number of items present in the array
print(a1.size)
print(a2.size)
print(a3.size)

#4. itemsize --> How much space occupies by each item of the array in the memory.
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)

#5. dtype --> Data type of items
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```
