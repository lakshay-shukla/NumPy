Array Operations

```python
# Array Operations
import numpy as np
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
print(a1)
print(a2)

# Scaler Operation --> Perform Single number operation on numpy array,
#1. Arithmetic  --> +,-,*,/,** etc.
print(a1 ** 2)

#2. Relational --> (<,>,>=,<=,==,!= etc.
print(a2==15)

# Vector Operations --> You can apply operation on 2 numpy Arrays. and array shape is same.
#1. Arithmetic
print(a1 * a2)
```

Changing Datatype

```python
# Changing Datatype
# astype
print(a1.astype(np.int32))
```

Array Functions

```python
# Array Functions
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
print(a1)

#1. max/min/sum/prod
# 0 -> col and 1 -> row
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))
# you can find max/min/sum/prod item in each row and column.
print(np.max(a1, axis=1))
print(np.min(a1, axis=1))
print(np.sum(a1, axis=0))
print(np.prod(a1, axis=0))

#2. mean/median/std/var
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
# you can find mean/median/std/var item in each row and column.
print(np.mean(a1, axis=0))
print(np.median(a1, axis=1))
print(np.std(a1, axis=0))
print(np.var(a1, axis=1))

#3. Trigonometric functions
print(np.sin(a1))

#4. Dot product --> Multiplication of 2 matrix
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)
print(np.dot(a2,a3))

#5. log and exponents
print(np.log(a1))
print(np.exp(a1))

#6. round/floor/ceil
print(np.round(np.random.random((2,3))*100))
print(np.floor(np.random.random((2,3))*100))
print(np.ceil(np.random.random((2,3))*100))
```
